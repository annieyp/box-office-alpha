```python
#data preparation and model training with XGBoost 
from data import prepare_data, split_data
from train import train_model

stocks = ['DIS', 'NFLX', 'WBD', 'PARA', 'SONY', 'LGF.A']

for stock in stocks:
    df = prepare_data(stock)
    X_train, X_test, y_train, y_test = split_data(df)
    preds = train_model(X_train, X_test, y_train, y_test, stock)
    
    plt.figure(figsize=(10, 5))
    plt.plot(y_test.values, label="Actual")
    plt.plot(preds, label="Predicted")
    plt.title(f"{stock} Stock Price Prediction")
    plt.xlabel("Days")
    plt.ylabel("Price")
    plt.legend()
    plt.grid()
    plt.show()

    accuracy = ( (preds[1:] > preds[:-1]) == (y_test.values[1:] > y_test.values[:-1]) ).mean()
    print(f"Directional Accuracy: {accuracy:.2%}")
```
